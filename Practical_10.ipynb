```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris(as_frame=True)
df = iris.frame
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))

# 1. Feature types
print("Feature Types:")
print(df.dtypes)

# 2. Histograms
df.drop(columns=['target', 'species']).hist(figsize=(10, 6), edgecolor='black')
plt.suptitle("Histograms of Iris Features")
plt.tight_layout()
plt.show()

# 3. Boxplots
df.drop(columns=['target', 'species']).plot(kind='box', subplots=True, layout=(2,2),
                                            figsize=(10,6), color='green', patch_artist=True)
plt.suptitle("Boxplots of Iris Features")
plt.tight_layout()
plt.show()

# 4. Boxplots by species (to compare and find outliers)
sns.pairplot(df, hue='species')
plt.suptitle("Pairwise Feature Distributions by Species", y=1.02)
plt.show()
```
