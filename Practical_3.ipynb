```python
from sklearn.datasets import load_iris
import numpy as np
```

```python
iris = load_iris(as_frame=True)
df_iris = iris.frame
```

```python
df_iris
```

```python
df_iris['species'] = iris.target
df_iris['species'] = df_iris['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
```

```python
for species in ['setosa', 'versicolor', 'virginica']:
    print(f"\n--- Statistics for {species} ---")
    subset = df_iris[df_iris['species'] == species].drop('species', axis=1)
    desc = subset.describe(percentiles=[.25, .5, .75])
    print(desc)
```
